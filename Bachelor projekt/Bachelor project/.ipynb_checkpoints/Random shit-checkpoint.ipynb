```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def prob_func(n):
    return 1 - np.exp(-19*(np.log(n)**4))
```

```python
x = np.linspace(-10,10,1000)

n_list = [0.4,0.5,0.55,0.6,0.65,0.7,0.75,0.8,0.85,0.9,0.95,0.99, 1.2, 1.3, 1.4, 1.5,1.6,1.7,1.8,1.9, 2]

func_list = []

for i in n_list:
    func_list.append(prob_func(i))
```

```python
plt.plot(n_list, func_list, ".")
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(x))
```

```python
x = np.linspace(-10 ,10,1000)
plt.plot(x, sigmoid(x))
```
