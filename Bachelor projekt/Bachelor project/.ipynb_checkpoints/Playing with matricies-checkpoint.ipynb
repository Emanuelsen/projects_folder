```python
import numpy as np 
import matplotlib.pyplot as plt
import sys
```

```python
a = np.array([[1, 1],
              [1, 2]])
```

```python
np.linalg.eig(a)
```

```python
A = np.array([[1,1,0], 
            [1,1,1],
            [0,0,1]])
```

```python
AT = np.transpose(A)
print(AT)
```

```python
ATA = np.matmul(AT,A)
print(ATA)
```

```python
np.linalg.eig(ATA)
```

```python
svd_A = np.linalg.svd(A)
```

```python
print(svd_A)
```

```python
print(svd_A[2])
print(svd_A[2][0])
```

```python
svd_A_1 = svd_A[1][0] * svd_A[2][0]
svd_A_2 = svd_A[1][1] * svd_A[2][1]
svd_A_3 = svd_A[1][2] * svd_A[2][2]
```

```python
fig = plt.figure()
ax = plt.axes(projection = "3d")
ax.set_xlim([-2,2])
ax.set_ylim([-2,2])
ax.set_zlim([-1,1])

start = [0,0,0]

ax.quiver(start[0], start[1], start[2], svd_A_1[0], svd_A_1[1], svd_A_1[2], color = "red")
ax.quiver(start[0], start[1], start[2], svd_A_2[0], svd_A_2[1], svd_A_2[2], color = "green")
ax.quiver(start[0], start[1], start[2], svd_A_3[0], svd_A_3[1], svd_A_3[2])
```
